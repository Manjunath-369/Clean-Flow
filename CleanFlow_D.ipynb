```python
import pandas as pd
from pathlib import Path

df_dwn_sum = pd.read_csv(r"C:\Users\manju\OneDrive\Documents\assign task automation texst case 1.csv")
dfd = pd.read_csv(r"C:\Users\manju\OneDrive\Documents\sid damage.csv").fillna("")
sheet_url = "https://docs.google.com/spreadsheets/d/1HpW2ggD9kxV9Ur3037mMpD-0f2GM7AFSTSuNVo2dEJY/export?format=csv&gid=0"
df_sheet = pd.read_csv(sheet_url)

dfd = dfd.sort_values("WTG #")
dfd["number_rs"] = dfd["WTG ID"].str.extract(r'(\d+)').fillna(0).astype(int)
dfd = dfd.sort_values(by="number_rs")

dfd = dfd.rename(columns={"number_rs": "num"})
dfd["Site Name"] = df_dwn_sum["Site"].iloc[0]
dfd["Site_Num"] = dfd["Site Name"].astype(str) + "_" + dfd["num"].astype(str)
dfd["Photo File Name"] = dfd["Photo File Name"].str.lower()

wtg_convert = df_dwn_sum.set_index("Site_Num")["Turbine"]
dfd["WTG #"] = dfd["Site_Num"].map(wtg_convert)

dfd["Component"] = dfd["Component"].str.title()
dfd["Material"] = dfd["Material"].str.title()
dfd["Type"] = dfd["Type"].str.title()

dfd["Chord"] = dfd["Chord"].clip(upper=100)
dfd["Date"] = pd.to_datetime(dfd["Date"])
dfd["Date"] = dfd["Date"].dt.strftime("%Y-%m-%d")
dfd.insert(22,"Inspection Date",dfd["Date"])
dfd = dfd.reset_index(drop=True)
dfd
# dfd.to_csv(fr"C:\Users\manju\Downloads\{site_name} {site_start} to {site_end} damage.csv", index=False)
```

```python
turbine_first = df_sheet["Turbine"].iloc[0]
turbine_last = df_sheet["Turbine"].iloc[-1]
df_site_name = df_sheet["Site"].iloc[0]
folder_name = f"{df_site_name} {turbine_first} to {turbine_last}"
desktop = Path(r"C:\Users\manju\OneDrive\Documents\Desktop\Skyspec")
output_folder = desktop / folder_name 

if output_folder.exists():
    print(f"Folder '{folder_name}' already exists, Delete that folder first")
else:
    output_folder.mkdir()
    print(f"Folder '{folder_name}' created successfully.")
dfd.to_csv(output_folder / "final_dmg.csv", index=False)  
```
